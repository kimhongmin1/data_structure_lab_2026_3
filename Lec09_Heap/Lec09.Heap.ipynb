```python
class BinHeap:
  def __init__(self):
    self.heap_list = [0]
    self.current_size = 0

  def perc_up(self, i):
    while i // 2 > 0:
      if self.heap_list[i] < self.heap_list[i//2]:
        tmp = self.heap_list[i//2]
        self.heap_list[i//2] = self.heap_list[i]
        self.heap_list[i] = tmp
      i = i // 2

  def insert(self, k):
    self.heap_list.append(k)
    self.current_size = self.current_size + 1
    self.perc_up(self.current_size)

  def perc_down(self, i):
    while (i * 2) <= self.current_size:
      mc = self.min_child(i)
      if self.heap_list[i] > self.heap_list[mc]:
        tmp = self.heap_list[i]
        self.heap_list[i] = self.heap_list[mc]
        self.heap_list[mc] = tmp
      i = mc

  def min_child(self, i):
    if i * 2 + 1 > self.current_size:
      return i * 2
    else:
      if self.heap_list[i*2] < self.heap_list[i*2+1]:
        return i*2
      else:
        return i*2+1

  def del_min(self):
    retval = self.heap_list[1]
    self.heap_list[1] = self.heap_list[self.current_size]
    self.current_size = self.current_size - 1
    self.heap_list.pop()
    self.perc_down(1)
    return retval # Corrected typo: ret_val -> retval

  def build_heap(self, alist):
    i = len(alist) // 2 # Corrected typo: a_list -> alist
    self.current_size = len(alist) # Corrected typo: a_list -> alist
    self.heap_list = [0] + alist[:] # Corrected typo: a_list -> alist
    while (i > 0):
      self.perc_down(i)
      i=i-1

  def heapsort(iterable):
    heap = BinHeap()
    for x in iterable:
      heap.insert(x)
    sorted_list = [] # Corrected typo: sirted_list -> sorted_list
    while heap.current_size > 0: # Corrected loop condition
      sorted_list.append(heap.del_min())
    return sorted_list
```

```python
my_heap = BinHeap()

# 리스트 정의
my_list = [9, 5, 6, 2, 3]

# build_heap 메서드를 사용하여 힙 생성 (리스트에서)
heap_from_list = BinHeap()
heap_from_list.build_heap(my_list)
print(f"build_heap으로 생성된 힙: {heap_from_list.heap_list[1:]}")

# 힙에 요소 삽입 (개별적으로)
my_heap.insert(9)
my_heap.insert(5)
my_heap.insert(6)
my_heap.insert(2)
my_heap.insert(3)
print(f"삽입 후 힙 리스트: {my_heap.heap_list[1:]}")

# 최소값 삭제
min_val = my_heap.del_min()
print(f"삭제된 최소값: {min_val}")
print(f"최소값 삭제 후 힙 리스트: {my_heap.heap_list[1:]}")

# heapsort 함수를 사용하여 리스트 정렬
sorted_result = BinHeap.heapsort([9, 5, 6, 2, 3])
print(f"heapsort 결과: {sorted_result}")

# 더 큰 리스트로 heapsort 테스트
large_list = [34, 1, 67, 8, 23, 90, 5, 12, 45, 78]
sorted_large_list = BinHeap.heapsort(large_list)
print(f"큰 리스트 heapsort 결과: {sorted_large_list}")
```
